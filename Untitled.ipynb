```python
from datasets import load_dataset

print("Loading Document-Type-Detection dataset...")

ds = load_dataset(
    "prithivMLmods/Document-Type-Detection",
    split="train"
)

print(ds)
print("\nFeatures:")
print(ds.features)

print("\nFirst sample:")
print(ds[0])
```

```python
from collections import Counter
import pandas as pd

# Get class names
class_names = ds.features["label"].names

# Count labels
counts = Counter(ds["label"])

distribution = pd.DataFrame({
    "Class ID": list(counts.keys()),
    "Class": [class_names[i] for i in counts.keys()],
    "Images": list(counts.values())
})

distribution = distribution.sort_values("Class ID").reset_index(drop=True)

print(distribution)
```

```python
import os
from collections import defaultdict
from PIL import Image
from tqdm.auto import tqdm

# Classes we selected
selected_classes = [
    "Hand-Written",
    "Invoice-Doc",
    "Letter-Doc",
    "Resume-Doc"
]

# Number of images per class
images_per_class = 1000

# Output directory
output_dir = "data/selected"

for class_name in selected_classes:
    os.makedirs(os.path.join(output_dir, class_name), exist_ok=True)

# Map class name -> label ID
class_to_id = {
    name: i for i, name in enumerate(ds.features["label"].names)
}

print("Selected classes:")
for class_name in selected_classes:
    print(f"{class_name} -> label {class_to_id[class_name]}")
```

```python
from collections import defaultdict
from tqdm.auto import tqdm

selected_ids = {
    class_to_id[name]: name
    for name in selected_classes
}

class_counts = defaultdict(int)

print("Saving selected images...")

for idx in tqdm(range(len(ds))):
    label = ds[idx]["label"]

    if label not in selected_ids:
        continue

    class_name = selected_ids[label]

    if class_counts[class_name] >= images_per_class:
        continue

    image = ds[idx]["image"]

    # Save as JPEG
    filename = f"{class_name}_{class_counts[class_name]:04d}.jpg"
    filepath = os.path.join(output_dir, class_name, filename)

    image.convert("RGB").save(filepath, "JPEG", quality=90)

    class_counts[class_name] += 1

    # Stop once all classes have enough images
    if all(class_counts[c] >= images_per_class for c in selected_classes):
        break

print("\nDataset extraction complete!")

for class_name in selected_classes:
    print(f"{class_name}: {class_counts[class_name]} images")
```

```python
import os
import shutil
import random

# Reproducibility
random.seed(42)

source_dir = "data/selected"
base_dir = "data"

classes = [
    "Hand-Written",
    "Invoice-Doc",
    "Letter-Doc",
    "Resume-Doc"
]

# Split configuration
splits = {
    "train": 0.70,
    "val": 0.15,
    "test": 0.15
}

# Create directories
for split in splits:
    for class_name in classes:
        os.makedirs(
            os.path.join(base_dir, split, class_name),
            exist_ok=True
        )

# Split images
for class_name in classes:

    class_path = os.path.join(source_dir, class_name)

    images = [
        f for f in os.listdir(class_path)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    random.shuffle(images)

    n = len(images)

    train_end = int(n * splits["train"])
    val_end = train_end + int(n * splits["val"])

    split_files = {
        "train": images[:train_end],
        "val": images[train_end:val_end],
        "test": images[val_end:]
    }

    for split, files in split_files.items():

        destination = os.path.join(
            base_dir,
            split,
            class_name
        )

        for filename in files:
            src = os.path.join(class_path, filename)
            dst = os.path.join(destination, filename)

            shutil.copy2(src, dst)

print("Dataset split completed successfully!")
```

```python
from collections import Counter
import os

for split in ["train", "val", "test"]:

    print(f"\n{split.upper()}")

    total = 0

    for class_name in classes:

        folder = os.path.join(
            "data",
            split,
            class_name
        )

        count = len([
            f for f in os.listdir(folder)
            if f.lower().endswith((".jpg", ".jpeg", ".png"))
        ])

        print(f"{class_name}: {count}")
        total += count

    print("Total:", total)
```

```python
import matplotlib.pyplot as plt
from PIL import Image
import os
import random

fig, axes = plt.subplots(2, 4, figsize=(16, 8))

for row, class_name in enumerate(classes[:2]):
    folder = os.path.join("data", "train", class_name)
    files = random.sample(os.listdir(folder), 4)

    for col, filename in enumerate(files):
        image = Image.open(os.path.join(folder, filename))

        axes[row, col].imshow(image)
        axes[row, col].set_title(class_name)
        axes[row, col].axis("off")

plt.tight_layout()
plt.show()
```
